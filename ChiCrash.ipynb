```python
import pandas as pd

df = pd.read_parquet(
    "s3://chitown-crashes-api/clean/crashes.parquet"
)

print(df.shape)
df.head()
```

```python
sample_df = df.sample(n=10000, random_state=42)

print(sample_df.shape)
sample_df.head()
```

```python
print(sample_df.columns.tolist())
```

```python
print(df["is_fatal"].value_counts())

print(df["is_fatal"].value_counts(normalize=True) * 100)
```

```python
fatal = df[df["is_fatal"] == 1]

non_fatal = df[df["is_fatal"] == 0].sample(
    n=895 * 5,
    random_state=42
)

sample_df = pd.concat([fatal, non_fatal])

# Shuffle the rows
sample_df = sample_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print(sample_df["is_fatal"].value_counts())
print(sample_df.shape)
```

```python
features = [
    "posted_speed_limit",
    "weather_condition",
    "lighting_condition",
    "first_crash_type",
    "trafficway_type",
    "roadway_surface_cond",
    "prim_contributory_cause",
    "crash_hour",
    "crash_day_of_week",
    "num_units",
    "intersection_related_i",
    "hit_and_run_i",
    "any_impaired",
    "any_unbelted",
    "any_pedestrian",
    "any_bicycle",
    "driver_age"
]

X = sample_df[features]
y = sample_df["is_fatal"]

print(X.shape)
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
print(X.shape)
print(y.shape)
```

```python
print(X_train.isnull().sum().sort_values(ascending=False))
```

```python
X_train = X_train.copy()
X_test = X_test.copy()

# Fill missing Yes/No values
X_train["intersection_related_i"] = X_train["intersection_related_i"].fillna("N")
X_test["intersection_related_i"] = X_test["intersection_related_i"].fillna("N")

X_train["hit_and_run_i"] = X_train["hit_and_run_i"].fillna("N")
X_test["hit_and_run_i"] = X_test["hit_and_run_i"].fillna("N")

# Fill missing driver ages with training median
median_age = X_train["driver_age"].median()

X_train["driver_age"] = X_train["driver_age"].fillna(median_age)
X_test["driver_age"] = X_test["driver_age"].fillna(median_age)

print(X_train.isnull().sum())
```

```python
X_train = pd.get_dummies(X_train)
X_test = pd.get_dummies(X_test)

# Make sure train and test have identical columns
X_train, X_test = X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model trained!")
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np

# Get importance from each model
rf_importance = model.feature_importances_
gb_importance = gb_model.feature_importances_

# Logistic regression uses coefficients
log_importance = np.abs(log_model.coef_[0])

importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Random_Forest": rf_importance,
    "Logistic_Regression": log_importance,
    "Gradient_Boosting": gb_importance
})

# Normalize each model's importance to 0-1
for col in ["Random_Forest", "Logistic_Regression", "Gradient_Boosting"]:
    importance_df[col] = (
        importance_df[col] / importance_df[col].max()
    )

# Average all three models
importance_df["Average_Importance"] = importance_df[
    ["Random_Forest", "Logistic_Regression", "Gradient_Boosting"]
].mean(axis=1)

top_10 = importance_df.sort_values(
    "Average_Importance",
    ascending=False
).head(10)

top_10
```

```python
def get_original_feature(feature):
    original_features = [
        "posted_speed_limit",
        "weather_condition",
        "lighting_condition",
        "first_crash_type",
        "trafficway_type",
        "roadway_surface_cond",
        "prim_contributory_cause",
        "crash_hour",
        "crash_day_of_week",
        "num_units",
        "intersection_related_i",
        "hit_and_run_i",
        "any_impaired",
        "any_unbelted",
        "any_pedestrian",
        "any_bicycle",
        "driver_age"
    ]

    for original in original_features:
        if feature == original or feature.startswith(original + "_"):
            return original

    return feature


importance_df["Original_Feature"] = (
    importance_df["Feature"].apply(get_original_feature)
)

grouped_importance = (
    importance_df
    .groupby("Original_Feature")[
        ["Random_Forest", "Logistic_Regression", "Gradient_Boosting"]
    ]
    .sum()
)

# Normalize again after grouping
for col in ["Random_Forest", "Logistic_Regression", "Gradient_Boosting"]:
    grouped_importance[col] = (
        grouped_importance[col] / grouped_importance[col].max()
    )

grouped_importance["Average_Importance"] = grouped_importance[
    ["Random_Forest", "Logistic_Regression", "Gradient_Boosting"]
].mean(axis=1)

top_10_factors = grouped_importance.sort_values(
    "Average_Importance",
    ascending=False
).head(10)

top_10_factors
```

```python
factors = [
    "first_crash_type",
    "prim_contributory_cause",
    "weather_condition",
    "lighting_condition",
    "trafficway_type",
    "roadway_surface_cond",
    "any_impaired",
    "any_unbelted",
    "any_pedestrian",
    "any_bicycle",
    "hit_and_run_i",
    "intersection_related_i"
]

results = []

for factor in factors:
    stats = df.groupby(factor)["is_fatal"].agg(
        ["count", "sum", "mean"]
    ).reset_index()

    stats.columns = [
        "Condition",
        "Total_Crashes",
        "Fatal_Crashes",
        "Fatal_Rate"
    ]

    stats["Factor"] = factor
    stats["Fatal_Rate"] = stats["Fatal_Rate"] * 100

    results.append(stats)

fatal_factors = pd.concat(results, ignore_index=True)

fatal_factors.head()
```

```python
reliable_factors = fatal_factors[
    fatal_factors["Total_Crashes"] >= 100
]
```

```python
overall_fatal_rate = df["is_fatal"].mean() * 100

print("Overall fatal crash rate:", overall_fatal_rate)
```

```python
reliable_factors = reliable_factors.copy()

reliable_factors["Risk_Ratio"] = (
    reliable_factors["Fatal_Rate"] /
    overall_fatal_rate
)
```

```python
top_10_risk = reliable_factors.sort_values(
    "Risk_Ratio",
    ascending=False
).head(10)

top_10_risk[
    [
        "Factor",
        "Condition",
        "Total_Crashes",
        "Fatal_Crashes",
        "Fatal_Rate",
        "Risk_Ratio"
    ]
]
```
